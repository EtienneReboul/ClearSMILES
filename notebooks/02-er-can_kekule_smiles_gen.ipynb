```python
from rdkit import Chem
import pandas as pd
import multiprocessing as mp
```

```python
moses_df = pd.read_csv("../data/raw/whole_original_MOSES.csv")
```

```python
def smiles2can_kekule(smiles : str) -> str:
    """
    Convert a SMILES string to a Kekulé canonical SMILES.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None
    kekule_smiles = Chem.MolToSmiles(mol, kekuleSmiles=True, canonical=True)
    return kekule_smiles
```

```python
moses_df
```

```python

with mp.Pool(mp.cpu_count()) as pool:
    results = pool.map(smiles2can_kekule, moses_df["SMILES"])
```

```python
can_kekule_df = pd.DataFrame(results, columns=["smiles"])
can_kekule_df["SPLIT"] = moses_df["SPLIT"]
```

```python
train_df = can_kekule_df[can_kekule_df["SPLIT"] == "train"]
train_df.drop(columns=["SPLIT"], inplace=True)
train_df.to_csv("../data/processed/moses_train_can_kekule.csv", index=False)
```

```python
test_df = can_kekule_df[can_kekule_df["SPLIT"] == "test"]
test_df.drop(columns=["SPLIT"], inplace=True)
test_df.to_csv("../data/processed/moses_test_can_kekule.csv", index=False)
```

```python
test_scaffold_df = can_kekule_df[can_kekule_df["SPLIT"] == "test_scaffolds"]
test_scaffold_df.drop(columns=["SPLIT"], inplace=True)
test_scaffold_df.to_csv("../data/processed/moses_test_scaffolds_can_kekule.csv", index=False)
```
